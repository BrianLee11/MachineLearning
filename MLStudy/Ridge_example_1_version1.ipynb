```python
# Step 0. import necessary libraries
import pandas as pd
import numpy as np

# for machine learning
from sklearn.linear_model import LinearRegression, Ridge, Lasso
```

```python
# Step 1. Data loading
# we use boston dataset
data_url = "http://lib.stat.cmu.edu/datasets/boston"
raw_df = pd.read_csv(data_url, sep="\s+", skiprows=22, header=None)

# Combine every two rows into one full data row
data = np.hstack([
    raw_df.values[::2, :],     # even rows: 3 values
    raw_df.values[1::2, :2]    # odd rows: only first 2 values
])

# Extract the target value (3rd column of odd-numbered rows)
target = raw_df.values[1::2, 2]
```

```python
print(data.shape) # ouput: (506,13)
# number of observations: 506
# number of dimensions in one observation: 13
# therefore, each input has 13 features
```

```python
print(target.shape) # output: (506, )
# number of observations: 506
```

```python
# data naming
X = data
y = target
```

```python
# Step 2. Create instance of Linear Regression()
lr1 = LinearRegression()
```

```python
# use the fit method to learn linear relationship of X and y
lr1.fit(X,y)
```

```python
# we can use predict([[13 inputs]])
predicted_target = lr1.predict([[1]*13])
print("Predicted target:", predicted_target[0])
```

```python
# source from the website
# url: https://lib.stat.cmu.edu/datasets/boston
feature_names = [
    'CRIM',    # per capita crime rate by town
    'ZN',      # proportion of residential land zoned for lots over 25,000 sq.ft.
    'INDUS',   # proportion of non-retail business acres per town
    'CHAS',    # Charles River dummy variable (= 1 if tract bounds river; 0 otherwise)
    'NOX',     # nitric oxides concentration (parts per 10 million)
    'RM',      # average number of rooms per dwelling
    'AGE',     # proportion of owner-occupied units built prior to 1940
    'DIS',     # weighted distances to five Boston employment centres
    'RAD',     # index of accessibility to radial highways
    'TAX',     # full-value property-tax rate per $10,000
    'PTRATIO', # pupil-teacher ratio by town
    'B',       # 1000(Bk - 0.63)^2 where Bk is the proportion of Black people by town
    'LSTAT'    # % lower status of the population
]
```

```python
print(lr1.coef_)        # gives estimated coefficients
print(lr1.intercept_)   # gives estimated intercept term
print(lr1.coef_.shape)  # Output: (13,)
```

## Result 1: using regular OLS

```python
# result 1 using regular OLS (ordinary least squares)
print("Linear Regression")
for f, w in zip(feature_names, lr1.coef_):
    print("{0:7s}: {1:6.2f}".format(f, w))
print("Linear Regression's Loss function: sum of squared coefficients = {0:4.2f}".format(sum(lr1.coef_**2)))
```

## Result 2: using Ridge

```python
lr2 = Ridge(alpha = 10.0) # explain alpha: adds L2 regularization to the loss function. This means it penalizes large weights (coefficients).
```

## 🧠 What is `Ridge(alpha=10.0)`?

- `Ridge` adds **L2 regularization** to the loss function. This means it penalizes large weights (coefficients).
- **Loss function for Ridge**:

  $\text{Loss} = \text{RSS} + \alpha \sum w_j^2$

  where:
  - **RSS** = Residual Sum of Squares
  - $\sum w_j^2$ = sum of squares of the model coefficients
  - `alpha` (λ) controls the **strength of the penalty**

---

## What does `alpha=10.0` do?

- Adds **moderate regularization**
- **Higher** `alpha` → more shrinkage (coefficients pulled toward 0)
- **Lower** `alpha` → behaves more like ordinary least squares (`LinearRegression`)

```python
lr2.fit(X,y)
print("Ridge")
for f,w in zip(feature_names, lr2.coef_):
    print("{0:7s}: {1:6.2f}".format(f, w))
print("Ridge's Loss function: sum of squared coefficients = {0:4.2f}".format(sum(lr2.coef_**2)))
```

## Result 3: using Lasso

```python
# LASSO
lr3 = Lasso(alpha = 2.0)
lr3.fit(X, y)
print("Lasso")
for f,w in zip(feature_names, lr3.coef_):
    print("{0:7s}: {1:6.2f}".format(f, w))
print("Lasso's Loss function: sum of squared coefficients = {0:4.2f}".format(sum(lr3.coef_**2)))
```

```python
# Comparison of three measurement of linear regression (ordinary), ridge, lasso
print("Linear Regression's Loss function: sum of squared coefficients = {0:4.2f}".format(sum(lr1.coef_**2)))
print("Ridge's Loss function: sum of squared coefficients = {0:4.2f}".format(sum(lr2.coef_**2)))
print("Lasso's Loss function: sum of squared coefficients = {0:4.2f}".format(sum(lr3.coef_**2)))
```

```python
# Comparing estimated coefficients from 3 models: lr1 (Linear), lr2 (Ridge), lr3 (e.g., Lasso)
for f, w1, w2, w3 in zip(feature_names, lr1.coef_, lr2.coef_, lr3.coef_):
    print("{0:7s}: {1:8.2f}  {2:8.2f}  {3:8.2f}".format(f, w1, w2, w3))
```
